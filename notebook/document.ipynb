```python
from pathlib import Path

from langchain_community.document_loaders import (
    DirectoryLoader,
    PyMuPDFLoader,
    UnstructuredImageLoader,
)


def load_documents(data_dir: str):

    data_path = Path(data_dir)

    # --------------------------------------------------------
    # Load PDFs
    # --------------------------------------------------------

    pdf_loader = DirectoryLoader(
        str(data_path),
        glob="**/*.pdf",
        loader_cls=PyMuPDFLoader,
        show_progress=True,
    )

    documents = pdf_loader.load()

    # --------------------------------------------------------
    # Load images
    # --------------------------------------------------------

    image_extensions = [
        "*.png",
        "*.jpg",
        "*.jpeg",
        "*.tiff",
        "*.tif",
        "*.bmp",
        "*.webp",
    ]

    for extension in image_extensions:

        image_loader = DirectoryLoader(
            str(data_path),
            glob=f"**/{extension}",
            loader_cls=UnstructuredImageLoader,
            loader_kwargs={
                "mode": "elements",
                "strategy": "fast",
            },
            show_progress=True,
        )

        documents.extend(image_loader.load())

    return documents


# ============================================================
# Load everything
# ============================================================

documents = load_documents("../data/PDFLearn")


# ============================================================
# Summary
# ============================================================

print(f"Total documents loaded: {len(documents)}")


# ============================================================
# Inspect documents
# ============================================================

for i, doc in enumerate(documents):

    print("\n" + "=" * 80)
    print(f"Document {i + 1}")
    print("=" * 80)

    print("Source:", doc.metadata.get("source"))
    print("Page:", doc.metadata.get("page"))
    print("Content:", doc.page_content[:500])
```

```python
from pathlib import Path

from langchain_community.document_loaders import (
    DirectoryLoader,
    PyMuPDFLoader,
    UnstructuredImageLoader,
)


# ============================================================
# Configuration
# ============================================================

DATA_DIR = Path("../data/PDFLearn")


# ============================================================
# Load PDFs and Images
# ============================================================

def load_documents(data_dir: str):

    data_path = Path(data_dir)

    # --------------------------------------------------------
    # 1. Load PDF files
    # --------------------------------------------------------

    pdf_loader = DirectoryLoader(
        str(data_path),
        glob="**/*.pdf",
        loader_cls=PyMuPDFLoader,
        show_progress=True,
    )

    documents = pdf_loader.load()

    # --------------------------------------------------------
    # 2. Load image files
    # --------------------------------------------------------

    image_extensions = [
        "*.png",
        "*.jpg",
        "*.jpeg",
        "*.tiff",
        "*.tif",
        "*.bmp",
        "*.webp",
    ]

    for extension in image_extensions:

        image_loader = DirectoryLoader(
            str(data_path),
            glob=f"**/{extension}",
            loader_cls=UnstructuredImageLoader,
            loader_kwargs={
                "mode": "elements",
            },
            show_progress=True,
        )

        documents.extend(image_loader.load())

    return documents


# ============================================================
# Load all documents
# ============================================================

documents = load_documents("../data/PDFLearn")


# ============================================================
# Summary
# ============================================================

print("\n" + "=" * 70)
print("DOCUMENT LOADING SUMMARY")
print("=" * 70)

print(f"Total documents loaded: {len(documents)}")


# ============================================================
# Count PDFs and images
# ============================================================

pdf_count = 0
image_count = 0

for document in documents:

    source = document.metadata.get("source", "").lower()

    if source.endswith(".pdf"):
        pdf_count += 1

    elif source.endswith(
        (
            ".png",
            ".jpg",
            ".jpeg",
            ".tiff",
            ".tif",
            ".bmp",
            ".webp",
        )
    ):
        image_count += 1


print(f"PDF documents loaded  : {pdf_count}")
print(f"Image documents loaded: {image_count}")


# ============================================================
# Inspect loaded documents
# ============================================================

for i, document in enumerate(documents):

    print("\n" + "=" * 70)
    print(f"DOCUMENT {i + 1}")
    print("=" * 70)

    print("Source:")
    print(document.metadata.get("source"))

    print("\nPage:")
    print(document.metadata.get("page"))

    print("\nContent:")
    print(document.page_content[:1000])
```
